```python
import math
synthetic = [{"step": i, "value": round(math.sin(i / 3) * 10, 3)} for i in range(24)]
print("SYNTHETIC rows:", len(synthetic))
```

```python
%%sql
SELECT step % 4 AS bucket, round(sum(value), 3)::DOUBLE AS total FROM synthetic GROUP BY 1 ORDER BY 1
```
